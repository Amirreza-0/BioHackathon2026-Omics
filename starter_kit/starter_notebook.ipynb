## Step 4 - Your first model: classify by similarityCompare a new fingerprint to each class profile and pick the closest one. The similaritymeasure is cosine similarity: it asks whether two fingerprints point in the samedirection, ignoring how strong they are.This is not a toy. It is essentially how the Connectivity Map works, and it is one of thetwo official baselines.

```python
def unit(M):    """Scale every row to length 1 so we can compare directions."""    M = np.asarray(M, dtype=float)    return M / (np.linalg.norm(M, axis=1, keepdims=True) + 1e-9)def classify(fingerprints, profiles):    """Return a similarity score of every fingerprint to every class."""    sim = unit(fingerprints[profiles.columns]) @ unit(profiles).T    return pd.DataFrame(sim, index=fingerprints.index, columns=profiles.index)scores = classify(X_train, class_profiles)predicted = scores.idxmax(axis=1)print("first few predictions vs truth:")pd.DataFrame({"predicted": predicted, "true": y_train}).head(10)
```

## Step 5 - How good is it, really?The numbers above are dishonest: we scored the model on the same data we built it from.Two rules make the evaluation honest, and they are the rules the judges care about:1. **Never test on data you trained on.** Use the `cv_fold` column: train on folds 1-3,   test on fold 0, and so on.2. **Keep a drug together.** All doses of one drug are in the same fold already, because   otherwise the model just recognises the drug instead of the mechanism.

```python
from sklearn.metrics import f1_scorehonest_predictions = pd.Series(index=X_train.index, dtype=object)for fold in sorted(meta["cv_fold"].unique()):    held_out = meta["cv_fold"] == fold    profiles = X_train[~held_out].groupby(y_train[~held_out]).mean()    honest_predictions[held_out] = classify(X_train[held_out], profiles).idxmax(axis=1)# score per DRUG, not per row: average the doses firstdrug_truth = y_train.groupby(meta["compound"]).first()drug_pred  = honest_predictions.groupby(meta["compound"]).agg(lambda s: s.mode()[0])macro_f1 = f1_score(drug_truth, drug_pred.loc[drug_truth.index], average="macro")accuracy = (drug_pred.loc[drug_truth.index] == drug_truth).mean()print(f"macro-F1 : {macro_f1:.3f}   <- the official metric")print(f"accuracy : {accuracy:.3f}")
```

**macro-F1** averages the score of each class separately, so a rare class counts as muchas a common one. Guessing the biggest class every time gives a terrible macro-F1, which isthe point.Which classes does it get wrong? This table is often the most interesting result of thewhole day:

```python
confusion = pd.crosstab(drug_truth, drug_pred.loc[drug_truth.index],                        rownames=["true"], colnames=["predicted"])confusion
```

**Look at the off-diagonal entries and ask why.** Some confusions are modelling noise.Some are real biology: two targets in the same pathway genuinely do similar things to acell, and no amount of machine learning will separate them cleanly.
